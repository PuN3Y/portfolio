# Analyzing Used Car Listings on eBay Kleinanzeigen
We will be working on a dataset of used cars from eBay Kleinanzeigen, a [classifieds](https://en.wikipedia.org/wiki/Classified_advertising) section of the German eBay website. 

The dataset was originally scraped and uploaded to [Kaggle](https://www.kaggle.com/orgesleka/used-cars-database/data). The dataset has 50,000 data points, and was modified by Dataquest to emulate scraped data with unclean data entries.

The data dictionary provided with data is as follows:

- `dateCrawled` - When this ad was first crawled. All field-values are taken from this date.
- `name` - Name of the car.
- `seller` - Whether the seller is private or a dealer.
- `offerType` - The type of listing
- `price` - The price on the ad to sell the car.
- `abtest` - Whether the listing is included in an A/B test.
- `vehicleType` - The vehicle Type.
- `yearOfRegistration` - The year in which the car was first registered.
- `gearbox` - The transmission type.
- `powerPS` - The power of the car in PS.
- `model` - The car model name.
- `kilometer` - How many kilometers the car has driven.
- `monthOfRegistration` - The month in which the car was first registered.
- `fuelType` - What type of fuel the car uses.
- `brand` - The brand of the car.
- `notRepairedDamage` - If the car has a damage which is not yet repaired.
- `dateCreated` - The date on which the eBay listing was created.
- `nrOfPictures` - The number of pictures in the ad.
- `postalCode` - The postal code for the location of the vehicle.
- `lastSeenOnline` - When the crawler saw this ad last online.

The aim of this project is to clean the data and analyze the included used car listings.

```python
# We will import the libraries we will be using in our cleaning and analysis processes.
import pandas as pd
import numpy as np

#We will next open the 'auto.csv' file, and determine the type of encoding it uses.
autos = pd.read_csv('autos.csv', encoding = 'Latin-1')
autos.info()
autos.head()
```

We can see there are 20 columns in our dataset, most of which are `object` Dtypes - stored as strings. There are some columns, such as `vehicleType`, which contain null values; `notRepairedDamage` has the most, with 9829 null values.

To first start our cleaning process, we'll clean the names of the columns.
## Cleaning Columns

```python
autos.columns
```

We will make the following changes:
- Change the columns from camelcase to snakecase
- Rename some columns, to make them more accurate descriptions

```python
autos.columns = ['date_crawled', 'name', 'seller', 'offer_type', 'price', 'ab_test',
       'vehicle_type', 'registration_year', 'gearbox', 'power_ps', 'model',
       'odometer', 'registration_month', 'fuel_type', 'brand',
       'unrepaired_damage', 'ad_created', 'num_photos', 'postal_code',
       'last_seen']
autos.head()
```

The changes we have made make the data more readable and follow standard python convention for variable naming.
## Exploring and Cleaning
#### `seller`, `offer_type`, and `num_photos`
By exploring, we can find areas of the data we can easily clean.

```python
autos.describe(include='all')
```

We can immediately see there are some odd columns:
- Both `seller` and `offer_type` show only 2 unique data entries, with all but 1 being the same.
- The `num_photos` column also appears odd, as it is a numerical column with apparantly no difference in data between entries. We will investigate this further to verify.

```python
autos["num_photos"].value_counts()
```

We can see the value `0` is given in all the data fields. We can see that the three columns we mentioned provide no useful data for analysis, so we will start by removing them from the dataset.

```python
autos = autos.drop(["num_photos", "seller", "offer_type"], axis=1)
```

#### `price` and `odometer`
We can see that these columns are numerical values, which have extra characters - currency and distance measurements. We can remove them and rename the column to allow us to more easily mathematically manipulate the data.

```python
autos["price"] = (autos["price"]
                             .str.replace(",", "")
                             .str.replace("$", "")
                             .astype(int)
                 )

autos["odometer"] = (autos["odometer"]
                             .str.replace(",", "")
                             .str.replace("km", "")
                             .astype(int)
                    )

autos.rename({"odometer": "odometer_km"}, axis = 1, inplace = True)
autos[["price", "odometer_km"]].head()
```

## Exploring `price` and `odometer_km`
Having now cleaned the data, we can further explore the data and determine if we need to further remove/ clean the dataset.

```python
print("Shape:")
print(autos["price"].unique().shape)
print()
print("Describe:")
print(autos["price"].describe())
print()
print("Value Count:")
autos["price"].value_counts().head(15)
```

##### Prelimary Analysis
- We can see there are 2357 unique values.
- We can see we have many entries at a value of 0. We will most likely remove these data entries, since they only make a small proportion of the entire dataset.
- We have a maximum price of about $100,000,000. This is much too high to be a regular data entry, we will class it as an outlier. We will further explore the data and determine an appropriate cut-off for outliers, then remove them from the dataset as well.

```python
autos["price"].value_counts().sort_index(ascending=False).head(15)
```

From our analysis, we can see there are 14 cars in the dataset sold at extreme values of around `$1 million` or more. We will remove these from our dataset. We will also remove data entires with a price of `$0`. We will retain `$1` and above vehicles, as eBay is an auction site, so cars may genuinely be listed at `$1`.

```python
autos = autos[autos["price"].between(1, 351000)]
```

### Incorrect Entries for `registration_year`

```python
autos["registration_year"].describe()
```

This data should represent when the car was manufactured, however we can see there are outliers on either end. There are some cars registered as having been manufactured in 1000, and some in 9999. We have to deal with them. 

We immediately know any vehicle registering as beyond April 2016 is innacurate, as they would be from beyond when the listing was created. Determining the lower bounds is more challenging without further data. We can assume it would be in the early 1900s, when cars were being prodcued.

```python
autos = autos[autos["registration_year"].between(1900, 2016)]
autos = autos[ ~ ((autos["registration_year"] == 2016) & (autos["registration_month"] > 4))]
autos["registration_year"].value_counts(normalize = True).head(10)
```

We can see the most recent cars were only recently registered in the last 22 years.
## Exploring Price by Brand

```python
autos["brand"].value_counts(normalize = True)
```

We can see there are many brands out there, however many don't have many listings. We will limit our exploration to brands with at least a 5% market presence.

```python
brand_percentage = autos["brand"].value_counts(normalize = True)
brand_with_presence = brand_percentage[brand_percentage > 0.05].index
mean_prices = {}

for brand in brand_with_presence:
    branded_cars = autos[autos["brand"] == brand]
    mean_price = branded_cars["price"].mean()
    mean_prices[brand] = int(mean_price)

mean_prices
```

We can see there is a distinct price gap, in the most frequently listed brands:
- Audi, BMW and Mercedes Benz are more expensive
- Ford and Opel are less expensive
- Volkswagen is in between - this may explain its popularity, it may be a 'best of 'both worlds' option.

```python
bmp_series = pd.Series(mean_prices)
pd.DataFrame(bmp_series, columns=["mean_price"])
```

## Exploring Mileage

```python
mean_mileage = {}

for brand in brand_with_presence:
    branded_cars = autos[autos["brand"] == brand]
    mean_miles = branded_cars["odometer_km"].mean()
    mean_mileage[brand] = int(mean_miles)

bmm_series = pd.Series(mean_mileage)
pd.DataFrame(bmm_series, columns=["mean_mileage"])
```

```python
brand_mean_mileage = pd.Series(mean_mileage).sort_values(ascending=False)
brand_info = pd.DataFrame(brand_mean_mileage,columns=['mean_mileage'])

brand_mean_prices = pd.Series(mean_prices).sort_values(ascending=False)
brand_info["mean_price"] = brand_mean_prices 

brand_info
```

The range of car mileages does not vary as much as the prices do by brand. There is a slight trend to the more expensive vehicles having higher mileage, with the less expensive vehicles having lower mileage.

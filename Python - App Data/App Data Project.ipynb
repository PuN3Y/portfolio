We don't want any duplicate entries in the dataset, as it will skew the data analysis. For instance, in the Google Play Market dataset, we can see there are duplicates present for 'Instagram'.

```python
for app in google:
    name = app[0]
    if name == 'Instagram':
        print(app)
# for app in apple:
#     name = app[1]
#     if name == 'Mannequin Challenge' or name == 'VR Roller Coaster':
#         print(app)
```

In the above example, we can see there are 4 entries for Instagram present. Using the code below, we can see the number of duplicate entries in either dataset:

```python
#Apple
duplicates = []
uniques = []
for app in apple:
    name = app[1]
    if name in uniques:
        duplicates.append(name)
    else:
        uniques.append(name)
print('The number of Duplicate Entries in the Apple Dataset:', len(duplicates))
print('The number of Unique Entries in the Apple Dataset:', len(uniques))
#print(duplicates)
print("")

#Google
duplicates = []
uniques = []
for app in google:
    name = app[0]
    if name in uniques:
        duplicates.append(name)
    else:
        uniques.append(name)
print('The number of Duplicate Entries in the Google Dataset: ', len(duplicates))
print('The number of Unique Entries in the Google Dataset: ', len(uniques))
#print(duplicates)
```

We do not want to remove our data randomly. Instead we will use a specfic criterion to decide which data remains. Below is one of the data entries in question:

`['Instagram', 'SOCIAL', '4.5', '66577313', 'Varies with device', '1,000,000,000+', 'Free', '0', 'Teen', 'Social', 'July 31, 2018', 'Varies with device', 'Varies with device']`

In this, the *66577313* refers to the number of reviews the app has, at the moment it was scraped. As a result, we can assume the entry with the most number of reviews is the most recent. We will use this technique to determine which data to keep.

```python
reviews_max = {}
for app in google:
    name = app[0]
    n_reviews = float(app[3])
    if name in reviews_max and reviews_max[name] < n_reviews:
        reviews_max[name] = n_reviews
    elif name not in reviews_max:
        reviews_max[name] = n_reviews
#We only need to run this code for the google dataset, as we know there is no duplicate entries present in the apple dataset
#We know there were 9659 unique entires from before, and we want ot make sure that there is 9659 entries in the dictionary.
print(len(reviews_max))
```

As seen, we got a 9659 total values in the table, which is what we were aiming for. We can now use `reviews_max` to create a new dataset, but with clean entries. This is seen below:

```python
google_clean = []
already_added = []
for app in google:
    name = app[0]
    n_reviews = float(app[3])
    if (name not in already_added) and (reviews_max[name] == n_reviews):
        google_clean.append(app)
        already_added.append(name)
#This is to check we have got the correct length
explore_data(google_clean, 0, 2, True)
```

By adding all the clean data to a dictionary, which we use as a reference to create a new list, we now have the correct length of new clean data. By using the `explore_data` function we created earleir, we are able to to comb through our data and see if it is of the expected length.

By using the combined `if` statements earlier, we are then able to select only apps with the highest number of reviews and add them to the list.

Note: The following is a process to clean the apple data set using the same method as listed above.

```python
#Resetting Variables to allow us to clean the Apple Dataset.
reviews_max = {}
for app in apple:
    name = app[1]
    n_reviews = float(app[6])
    if name in reviews_max and reviews_max[name] < n_reviews:
        reviews_max[name] = n_reviews
    elif name not in reviews_max:
        reviews_max[name] = n_reviews
apple_clean = []
already_added = []
for app in apple:
    name = app[1]
    n_reviews = float(app[6])
    if (name not in already_added) and (reviews_max[name] == n_reviews):
        apple_clean.append(app)
        already_added.append(name)
#This is to check we have got the correct length
explore_data(apple_clean, 0, 2, True)
```

# Removing Non-English Applications

Our company focuses on producing apps for the englishs-speaking market, as a result, we aim to produce apps that will be popular in those markets; we need to remove non-english applications from both datasets.

In order to achieve our goals of scanning both databases, we can create a universal function that will check if the app is in English. Too limit the effect of data-loss, we will only remove the entry, if there are more than 3 non-ASCII english characters.

```python
def check_english(phrase):
    non_english_count = 0
    for char in phrase:
        if ord(char) > 127:
            non_english_count += 1
    #This count only allows the app to fail the check if there are more than 3 non-standard characters. This allows for non
    #standard characters in some English app titles, such as Emojis.
    if non_english_count > 3:
        return False
    return True
```

```python
#Apple (English Apps Only)
apple_english = []
for app in apple_clean:
    name = app[1]
    if check_english(name):
        apple_english.append(app)
#print(len(apple_english))
        
#Google (English Apps Only)
google_english = []
for app in google_clean:
    name = app[0]
    if check_english(name):
        google_english.append(app)
#print(len(google_english))

print('English Apple Dataset')
explore_data(apple_english, 0, 0, True)
print('')
print('English Google Dataset')
explore_data(google_english, 0, 0, True)
```

We can see in the new datset, the English-only component contains only 6181 entries in the Apple App Store, and 9614 entries in the Google Play Markets.

To minimize risks and overhead, our validation strategy for an app idea is comprised of three step:

Since our end goal is to produce an app that is suitable for both markets, our aim is to find a profile suitable for both demographics. We can start to do this by looking at the most commmon category of apps in either market. For this, we'll build a frequency table for the `prime_genre` column of the App Store data set, and the `Genres` and `Category` columns of the Google Play data set.

We can do this by creating frequency tables, we can use two different functions to analyze either frequency table to help us focus our aims:
- One function to generate frequency tables that show percentages
- Another function we can use to display the percentages in a descending order

```python
def freq_table(dataset, index):
    table = {}
    total = 0
    #This for loop will create a frequency table with numbers
    for row in dataset:
        total += 1
        value = row[index]
        if value in table:
            table[value] += 1
        else:
            table[value] = 1
 
    #This section of the code focuses on converting our raw data into percentages
    percentage_table = {}
    for key in table:
        percentage = (table[key] / total * 100)
        percentage_table[key] = percentage    
    return percentage_table

#<------------------------------------------------------>
#This code below was provided by DataQuest
def display_table(dataset, index):
    table = freq_table(dataset, index)
    table_display = []
    for key in table:
        key_val_as_tuple = (table[key], key)
        table_display.append(key_val_as_tuple)
        
    table_sorted = sorted(table_display, reverse = True)
    for entry in table_sorted:
        print(entry[1], ':', entry[0])
#<------------------------------------------------------>
```

# The Most Popular Apps on Google Play Markets by Genre

We can see in this dataset, there is `installs` data, however, as seen here, it can be very imprecise. There is no datapoint,in between values. We do not know if there is 1 million, or 1.2 million for an app which is 1,000,000+.

```python
display_table(final_google, 5)
```

In order to get around this issue, we will assume that for an app which is 1,000,000+, it has only 1 million installs and so on. We will do this for each checkpoint value. In order to manage this computationally, we must convert all the installation data entries into `floats`.

```python
categories_google = freq_table(final_google, 1)
google_categories_sorted = {}
for category in categories_google:
    total = 0
    len_category = 0
    for app in final_google:
        category_app = app[1]
        if category_app == category:
            n_installs = app[5]
            n_installs = n_installs.replace(',', '')
            n_installs = n_installs.replace('+', '')
            total += float(n_installs)
            len_category += 1
    avg_n_installs = total / len_category
    avg_n_installs = round(avg_n_installs,2)
    google_categories_sorted[category] = avg_n_installs
table_display = []
for key in google_categories_sorted:
    key_val_as_tuple = (google_categories_sorted[key], key)
    table_display.append(key_val_as_tuple)
    table_sorted = sorted(table_display, reverse = True)
    
for entry in table_sorted:
    print(entry[1], ':', entry[0])
```

We can see that communication contains apps, which have the most number of installs, however, this is a heavily skewed result. There are some apps with over a billion installs, with a few more with 100 million+:

```python
for app in final_google:
    if app[1] == 'COMMUNICATION' and (app[5] == '1,000,000,000+' or app[5] == '500,000,000+' or app[5] == '100,000,000+'):
        print(app[0], ':', app[5])
```
